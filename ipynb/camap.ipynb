```python
import pandas as pd
import gc
```

```python
df_full = pd.read_csv('Human_BLCL.scores.tsv.gz', sep='\t', index_col=[0,1,2,3,4], header=[0,1])
```

```python
df_full = df_full.loc[:, (slice(None), 'ScoresReduced')]
df_full.columns = df_full.columns.droplevel(1)
df_full.index = df_full.index.droplevel([0,1])
gc.collect()
```

```python
df = df_full.loc[(7, slice(None), slice(None)),:]
```

```python
df = df.reset_index().drop_duplicates().set_index(['N', 'Peptide', 'Class'])
```

```python
cs = df.reset_index().Class.sum()
ca = df.reset_index().Class.shape[0]
print(ca, cs)
```

```python
prefixes = [
    'CAMAPShuff',
    'CAMAP',
    'TPM',
    'TPM-CAMAPShuff',
    'TPM-CAMAP',
    'BS-CAMAPShuff',
    'BS-CAMAP',
    'BS-TPM-CAMAPShuff',
    'BS',
    'BS-TPM',
    'BS-TPM-CAMAP'
]
proportions = [
    0.01,
    0.1,
    0.25,
    0.5,
    0.75,
    0.9,
    0.99
]
```

```python
for proportion in proportions:
    print(proportion*100, '%', round(cs*proportion, 5))
    for prefix in prefixes:
        ccs = df[[prefix]].sort_values(prefix, ascending=False).reset_index().Class.cumsum()
        print(prefix, ':', ccs[ccs == int(cs*proportion)].index[0],
              round(ccs[ccs == int(cs*proportion)].index[0]/ca*100, 3), '%')
    print()
```
